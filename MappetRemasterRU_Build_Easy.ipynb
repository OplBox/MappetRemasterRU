# Сборка MappetRemasterRU и его документации

Этот ноутбук Google Colab поможет вам собрать проект MappetRemasterRU и его документацию. Просто вставьте ссылку на ваш репозиторий и запустите ячейки.

## 1. Клонирование репозитория

Вставьте ссылку на ваш GitHub репозиторий в поле ниже и запустите ячейку.

```python
repository_url = 'https://github.com/OplBox/MappetRemasterRU.git' #@param {type:"string"}
!git clone {repository_url}
%cd MappetRemasterRU
```

## 2. Установка зависимостей и сборка проекта

Эта ячейка установит все необходимые зависимости, соберет `jsonDoclet` и сам проект. Это может занять некоторое время.

```python
!sudo apt-get update
!sudo apt-get install -y openjdk-8-jdk maven
!git clone https://github.com/mchorse/jsonDoclet.git
%cd jsonDoclet
!mvn package
%cd ..
!mkdir -p run
!cp jsonDoclet/target/jsonDoclet-1.0-SNAPSHOT.jar run/jsonDoclet.jar
!chmod +x gradlew
!./gradlew build
!./gradlew scriptingJavadocs
```

## 3. Результаты

Сборка завершена! Вы можете найти результаты в следующих директориях:

*   **Собранный мод:** `build/libs/mappet-0.7.jar`
*   **Javadoc документация:** `build/docs/javadoc`
*   **JSON документация:** `src/main/resources/assets/mappet/docs.json`

Вы можете скачать эти файлы из файлового менеджера слева.
