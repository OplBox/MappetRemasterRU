## 2. Установка зависимостей и сборка проекта

```python
import os

# Установка и настройка Java 8
!sudo apt-get update
!sudo apt-get install -y openjdk-8-jdk
!sudo update-alternatives --set java /usr/lib/jvm/java-8-openjdk-amd64/jre/bin/java
!sudo update-alternatives --set javac /usr/lib/jvm/java-8-openjdk-amd64/bin/javac
os.environ['JAVA_HOME'] = '/usr/lib/jvm/java-8-openjdk-amd64'

# Сборка jsonDoclet
os.chdir('/content')
!rm -rf jsonDoclet
!git clone https://github.com/mchorse/jsonDoclet.git
os.chdir('/content/jsonDoclet')
!sed -i 's/5.2.0/2.0.4/' build.gradle
!chmod +x gradlew
!./gradlew build

# Сборка MappetRemasterRU
os.chdir('/content/MappetRemasterRU')
!mkdir -p run
!cp /content/jsonDoclet/build/libs/jsonDoclet-1.0.jar run/jsonDoclet.jar
!chmod +x gradlew

# --- ОТЛАДКА --- 
print("--- DEBUGGING START ---")
!echo "Current directory:" > /content/debug.txt
!pwd >> /content/debug.txt
!echo "
Directory listing for /content/MappetRemasterRU:" >> /content/debug.txt
!ls -la /content/MappetRemasterRU >> /content/debug.txt
!echo "
Directory listing for /content/MappetRemasterRU/gradle:" >> /content/debug.txt
!ls -la /content/MappetRemasterRU/gradle >> /content/debug.txt
!echo "
Directory listing for /content/MappetRemasterRU/gradle/wrapper:" >> /content/debug.txt
!ls -la /content/MappetRemasterRU/gradle/wrapper >> /content/debug.txt
!echo "
Content of gradle-wrapper.properties:" >> /content/debug.txt
!cat /content/MappetRemasterRU/gradle/wrapper/gradle-wrapper.properties >> /content/debug.txt
print("--- DEBUGGING OUTPUT ---")
!cat /content/debug.txt
print("--- DEBUGGING END ---")

!echo "
--- Starting MappetRemasterRU build... ---"
!./gradlew build
!./gradlew scriptingJavadocs
```
